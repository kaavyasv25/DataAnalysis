```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
superstore_df = pd.read_csv("SuperstoreSales.csv")  # Load the Superstore sales data
zillow_df = pd.read_csv("ZillowHomeValueIndex.csv")  # Load the Zillow Home Value Index data
```

```python
print(superstore_df.info())
print(superstore_df.head())
print(zillow_df.info())
print(zillow_df.head())
```

```python
superstore_df = superstore_df.drop(["Row_ID", "Country", "City", "Profit", "Payment_Mode", "Quantity", "Ship_Mode"], axis=1)
```

```python
superstore_df['Order_Date'] = pd.to_datetime(superstore_df['Order_Date'])
superstore_df['YearMonth'] = superstore_df['Order_Date'].dt.to_period('M')
```

```python
print(zillow_df.isnull().sum())
```

```python
state_abbr_to_full = {
    'AL': 'Alabama', 'AK': 'Alaska', 'AZ': 'Arizona', 'AR': 'Arkansas', 
    'CA': 'California', 'CO': 'Colorado', 'CT': 'Connecticut', 'DE': 'Delaware', 
    'FL': 'Florida', 'GA': 'Georgia', 'HI': 'Hawaii', 'ID': 'Idaho', 
    'IL': 'Illinois', 'IN': 'Indiana', 'IA': 'Iowa', 'KS': 'Kansas', 
    'KY': 'Kentucky', 'LA': 'Louisiana', 'ME': 'Maine', 'MD': 'Maryland', 
    'MA': 'Massachusetts', 'MI': 'Michigan', 'MN': 'Minnesota', 'MS': 'Mississippi', 
    'MO': 'Missouri', 'MT': 'Montana', 'NE': 'Nebraska', 'NV': 'Nevada', 
    'NH': 'New Hampshire', 'NJ': 'New Jersey', 'NM': 'New Mexico', 'NY': 'New York', 
    'NC': 'North Carolina', 'ND': 'North Dakota', 'OH': 'Ohio', 'OK': 'Oklahoma', 
    'OR': 'Oregon', 'PA': 'Pennsylvania', 'RI': 'Rhode Island', 'SC': 'South Carolina', 
    'SD': 'South Dakota', 'TN': 'Tennessee', 'TX': 'Texas', 'UT': 'Utah', 
    'VT': 'Vermont', 'VA': 'Virginia', 'WA': 'Washington', 'WV': 'West Virginia', 
    'WI': 'Wisconsin', 'WY': 'Wyoming'
}
zillow_df['StateName'] = zillow_df['StateName'].map(state_abbr_to_full)
```

```python
# Melt the DataFrame
zillow_long = zillow_df.melt(id_vars=['StateName'], var_name='Date', value_name='Home_Value_Index')

# Check the Date column format
print(zillow_long['Date'].head())  # Inspect first few rows of the Date column

# Specify the correct date format explicitly (assuming YYYY-MM-DD here, adjust if needed)
zillow_long['Date'] = pd.to_datetime(zillow_long['Date'], format='%Y-%m-%d', errors='coerce')

# Handle any invalid dates (NaT)
print(zillow_long[zillow_long['Date'].isna()])  # Check for invalid dates

# Extract YearMonth period
zillow_long['YearMonth'] = zillow_long['Date'].dt.to_period('M')
```

```python
zillow_long = zillow_long.dropna(subset=['Date', 'Home_Value_Index'])
```

```python
sales_by_state = superstore_df.groupby(['State', 'YearMonth'])['Sales'].sum().reset_index()
```

```python
merged_df = pd.merge(
    sales_by_state, 
    zillow_long, 
    left_on=['State', 'YearMonth'], 
    right_on=['StateName', 'YearMonth'], 
    how='inner'
)
```

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(
    data=merged_df, 
    x='Home_Value_Index', 
    y='Sales', 
    hue='State', 
    palette='tab20', 
    legend=None
)
plt.title("Relationship Between Home Value Index and Sales")
plt.xlabel("Home Value Index")
plt.ylabel("Sales")
plt.show()
```

```python
# Calculate correlation between sales and home value index
correlation = merged_df[['Sales', 'Home_Value_Index']].corr()
print("Correlation Matrix:")
print(correlation)

# Save merged dataset for further analysis
merged_df.to_csv("Merged_Superstore_Zillow.csv", index=False)
```

```python
# Group data by state and calculate correlation between Sales and Home Value Index
statewise_correlation = merged_df.groupby('State').apply(
    lambda x: x[['Sales', 'Home_Value_Index']].corr().iloc[0, 1]  # Correlation coefficient
).reset_index(name='Correlation')

import warnings
warnings.filterwarnings("ignore", category=UserWarning, module='seaborn')

# FacetGrid scatter plots for each state separately
g = sns.FacetGrid(merged_df, col="State", col_wrap=4, height=3)
g.map(sns.scatterplot, "Home_Value_Index", "Sales")
g.set_axis_labels("Home Value Index", "Sales")
g.set_titles("{col_name}")
plt.subplots_adjust(top=0.6)
g.fig.suptitle("Relationship Between Home Value Index and Sales by State", fontsize=16)
plt.show()
```
